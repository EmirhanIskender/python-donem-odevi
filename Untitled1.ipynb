# **1. KISIM: Tek müşteri bilgileri**

```python
# Rastgele sayı üretmek için random kütüphanesi
import random


# Müşteri bilgileri
ad_soyad = "Emirhan İskender"   # string veri tipi
aylik_ucret = 620.75            # float veri tipi
sadakat_puani = 28              # integer veri tipi
aktif_mi = True                 # boolean veri tipi


# Müşterinin kullandığı hizmetler listesi
hizmetler = [
    "Fiber İnternet",
    "TV+",
    "Bulut Depolama",
    "Mobil Hat",
    "Güvenlik Paketi"
]


# Müşteri bilgilerini sözlük içinde saklıyoruz
musteri = {

    "ad_soyad": ad_soyad,
    "aylik_ucret": aylik_ucret,
    "sadakat_puani": sadakat_puani,
    "aktif_mi": aktif_mi,
    "hizmetler": hizmetler
}


# Müşterinin VIP olup olmadığını kontrol eder
if musteri["aylik_ucret"] > 500 or musteri["sadakat_puani"] > 24:

    print("VIP Müşteri - Kampanya Tanımlandı")

else:

    print("Standart Müşteri")


# Müşteri adını büyük harfe çevirir
buyuk_harf_ad = musteri["ad_soyad"].upper()


# Rastgele müşteri numarası oluşturur
rastgele_id = random.randint(1000, 9999)


# Customer ID oluşturur
customer_id = f"CST-2026-{rastgele_id}"


# Sonuçları ekrana yazdırır
print("Müşteri Adı:", buyuk_harf_ad)

print("Customer ID:", customer_id)

print("Aylık Ücret:", musteri["aylik_ucret"])

print("Sadakat Puanı:", musteri["sadakat_puani"])

print("Alınan Hizmetler:", musteri["hizmetler"])
```

**Cevap :** Müşteri bilgilerini sözlük içinde saklamak daha avantajlıdır çünkü verilere anahtar isimleriyle kolayca ulaşılabilir. Ayrıca müşterinin aldığı hizmetler gibi bilgiler liste şeklinde düzenli olarak saklanabilir. Bu sayede kod daha okunabilir ve anlaşılır olur.

# **2. KISIM: Birden fazla müşteri**

```python
# Matematiksel işlemler için math kütüphanesi
import math

# Tarih ve saat işlemleri için datetime kütüphanesi
from datetime import datetime


# -----------------------------------
# 2. KISIM: Müşteri Analizi
# -----------------------------------


# Müşteri listesini oluşturur
musteriler_listesi = [

    # Her müşteri için sözlük yapısı kullanılmıştır
    {"ad": "Ali", "ucret": 400, "sadakat_ayi": 10, "churn": False},

    {"ad": "Ayşe", "ucret": 600, "sadakat_ayi": 30, "churn": False},

    {"ad": "Mehmet", "ucret": 300, "sadakat_ayi": 5, "churn": True},

    {"ad": "Fatma", "ucret": 750, "sadakat_ayi": 12, "churn": False},

    {"ad": "Can", "ucret": 450, "sadakat_ayi": 26, "churn": False}
]


# KDV hesaplayan fonksiyon
def tutar_hesapla(aylik_ucret):

    # Ücrete %20 KDV ekler
    kdv_dahil_tutar = aylik_ucret * 1.20

    # Hesaplanan değeri geri döndürür
    return kdv_dahil_tutar


# Başlık yazdırır
print("--- MÜŞTERİ FATURA ANALİZİ ---")


# Listedeki müşterileri tek tek dolaşır
for musteri in musteriler_listesi:

    # KDV dahil tutarı hesaplar
    net_tutar = tutar_hesapla(musteri["ucret"])

    # Sayıyı yukarı yuvarlar
    son_tutar = math.ceil(net_tutar)


    # Churn riski kontrolü yapar
    if musteri["sadakat_ayi"] < 6 or musteri["churn"] == True:

        durum = "Riskli"

    else:

        durum = "Güvenli"


    # Sonuçları ekrana yazdırır
    print(f"Müşteri: {musteri['ad']} | Fatura: {son_tutar} TL | Durum: {durum}")


# Güncel tarihi alır
bugun = datetime.now().strftime("%d/%m/%Y %H:%M:%S")


# Tarihi ekrana yazdırır
print(f"\nAnaliz Tarihi: {bugun}")


# Hizmet listesi oluşturur
tum_hizmetler = [
    "İnternet",
    "TV",
    "İnternet",
    "Telefon",
    "TV",
    "Bulut"
]


# Tekrar eden hizmetleri kaldırır
benzersiz_hizmetler = set(tum_hizmetler)


# Benzersiz hizmetleri yazdırır
print(f"Sunulan Benzersiz Hizmetler: {benzersiz_hizmetler}")
```

**2.Kısmın Sorusu :** Yazdığınız for döngüsü müşteri listesini gezerken churn riski taşıyan müşterileri nasıl tespit etti?

**Cevap :** For döngüsü müşteri listesindeki her müşteriyi tek tek kontrol eder. Sadakat ayı 6’dan küçük olan veya churn bilgisi True olan müşteriler riskli olarak değerlendirilmiştir.
